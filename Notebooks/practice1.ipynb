```python
y = df["Status"]

x = df.drop(columns=[
    "Status",
    "ID",
    "loan_limit",
    "Credit_Worthiness",
    "open_credit",
    "Security_Type"
]) #drop unique values
```

```python
num_cols = [
    "year",
    "Credit_Score",
    "LTV",
    "dtir1"
]
```

```python
binary_cols = [
    "approv_in_adv",
    "business_or_commercial",
    "submission_of_application"
]

binary_maps = {
    "approv_in_adv": {"pre": 1, "nopre": 0},
    "business_or_commercial": {"b/c": 1, "nob/c": 0},
    "submission_of_application": {"to_inst": 1, "not_inst": 0}
}

for col, mapping in binary_maps.items():
    x[col] = x[col].map(mapping)
```

```python
from sklearn.preprocessing import OrdinalEncoder

ordinal_cols = ["age"]

age_order = [["25-34", "35-44", "45-54", "55-64"]]

ordinal_encoder = OrdinalEncoder(
    categories = [["25-34", "35-44", "45-54", "55-64"]],
   handle_unknown="use_encoded_value",
   unknown_value=-1
)
```

```python
from sklearn.preprocessing import OneHotEncoder

nominal_cols = [
    "Gender",
    "loan_type",
    "loan_purpose",
    "credit_type",
    "co-applicant_credit_type",
    "Region"
]

onehot_encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)
```

```python
from sklearn.compose import ColumnTransformer

preprocessor = ColumnTransformer(
    transformers = [
        ("num", "passthrough", num_cols),
        ("bin", "passthrough", binary_cols),
        ("ord", ordinal_encoder, ordinal_cols),
        ("nom", onehot_encoder, nominal_cols)
    ]
)
```

```python
x.shape
type(x)
```

```python
x_encoded = preprocessor.fit_transform(x)

feature_names = preprocessor.get_feature_names_out()

x_encoded = pd.DataFrame(x_encoded, columns=feature_names)
```
